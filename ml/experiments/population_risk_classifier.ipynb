```python
import pandas as pd
import numpy as np
import joblib
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense

# 1. Load your 6 cleaned CSV files from Colab
rad = pd.read_csv('clean_1_radiation.csv')
gluc = pd.read_csv('clean_2_glucose.csv')
sleep_mood = pd.read_csv('clean_3_sleep_mood.csv')
muscle = pd.read_csv('clean_4_muscle.csv')
bone = pd.read_csv('clean_5_bone.csv')
vitals = pd.read_csv('clean_6_vitals.csv')

# 2. Sample 5,000 synchronized rows across all 6 cleaned sources
n = 5000
np.random.seed(42)

v_s = vitals.sample(n=n, replace=True, random_state=42).reset_index(drop=True)
g_s = gluc.sample(n=n, replace=True, random_state=42).reset_index(drop=True)
sm_s = sleep_mood.sample(n=n, replace=True, random_state=42).reset_index(drop=True)
m_s = muscle.sample(n=n, replace=True, random_state=42).reset_index(drop=True)
b_s = bone.sample(n=n, replace=True, random_state=42).reset_index(drop=True)
r_s = rad.sample(n=n, replace=True, random_state=42).reset_index(drop=True)

# 3. Combine into your strict 9-indicator master dataset
master_df = pd.DataFrame({
    'systolic_bp': v_s['systolic_bp'],
    'diastolic_bp': v_s['diastolic_bp'],
    'heart_rate': v_s['heart_rate'],
    'blood_sugar': g_s['blood_sugar'],
    'body_temperature': v_s['body_temperature'],
    'daily_sleep_hours': sm_s['daily_sleep_hours'],
    'mood': sm_s['mood'],
    'muscle_strength_kg': m_s['muscle_strength_kg'],
    'bone_density_t_score': b_s['bone_density_t_score'],
    'radiation_mgy_day': r_s['radiation_mgy_day']
})

# Inject 10% realistic spaceflight emergency spikes (solar flares, tachycardia, severe bone loss)
spike_idx = np.random.choice(n, size=int(n * 0.10), replace=False)
master_df.loc[spike_idx, 'radiation_mgy_day'] = np.random.uniform(0.45, 2.50, len(spike_idx)).round(3)
master_df.loc[spike_idx[:250], 'heart_rate'] = np.random.uniform(118, 148, 250).round(1)
master_df.loc[spike_idx[250:], 'bone_density_t_score'] = np.random.uniform(-2.7, -2.0, len(spike_idx) - 250).round(2)

# 4. Assign Spaceflight Risk Labels (0 = Optimal, 1 = Warning, 2 = Critical)
def assign_risk(row):
    if (
        row['systolic_bp'] > 145 or row['systolic_bp'] < 85 or
        row['heart_rate'] > 115 or row['heart_rate'] < 45 or
        row['blood_sugar'] < 65 or row['blood_sugar'] > 155 or
        row['body_temperature'] > 38.0 or row['body_temperature'] < 35.5 or
        row['bone_density_t_score'] < -2.0 or
        row['radiation_mgy_day'] > 0.40
    ):
        return 2  # Critical
    elif (
        row['systolic_bp'] > 130 or
        row['heart_rate'] > 95 or
        row['daily_sleep_hours'] < 6.0 or
        row['mood'] <= 2 or
        row['muscle_strength_kg'] < 28.0 or
        row['bone_density_t_score'] < -1.0 or
        row['radiation_mgy_day'] > 0.25
    ):
        return 1  # Warning
    return 0      # Optimal

master_df['risk_level'] = master_df.apply(assign_risk, axis=1)

# 5. Scale features & split data
X = master_df.drop(columns=['risk_level'])
y = master_df['risk_level']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# 6. Train Model A: Unsupervised Autoencoder (learns normal healthy baseline)
X_train_normal = X_train[y_train == 0]
input_dim = X_train.shape[1]

inp = Input(shape=(input_dim,))
enc = Dense(8, activation='relu')(inp)
bot = Dense(4, activation='relu')(enc)
dec = Dense(8, activation='relu')(bot)
out = Dense(input_dim, activation='linear')(dec)

autoencoder = Model(inp, out)
autoencoder.compile(optimizer='adam', loss='mse')
autoencoder.fit(X_train_normal, X_train_normal, epochs=40, batch_size=32, verbose=0)

train_recon = autoencoder.predict(X_train_normal, verbose=0)
threshold = np.percentile(np.mean(np.square(X_train_normal - train_recon), axis=1), 95)

# 7. Train Model B: Supervised Random Forest Classifier
rf_model = RandomForestClassifier(n_estimators=150, random_state=42)
rf_model.fit(X_train, y_train)

print("Autoencoder Anomaly Threshold (MSE):", round(threshold, 4))
print("\nRandom Forest Risk Classification Report:\n")
print(classification_report(y_test, rf_model.predict(X_test)))

# 8. Save all trained models & scaler for download
autoencoder.save('astronaut_autoencoder.keras')
joblib.dump(rf_model, 'astronaut_rf_model.pkl')
joblib.dump(scaler, 'astronaut_scaler.pkl')
joblib.dump({'threshold': threshold, 'features': list(X.columns)}, 'astronaut_config.pkl')
master_df.to_csv('final_astronaut_9features.csv', index=False)
print("Saved all models and final_astronaut_9features.csv successfully!")
import pandas as pd
import numpy as np
import joblib
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense

# 1. Load your 6 cleaned CSV files from Colab
rad = pd.read_csv('clean_1_radiation.csv')
gluc = pd.read_csv('clean_2_glucose.csv')
sleep_mood = pd.read_csv('clean_3_sleep_mood.csv')
muscle = pd.read_csv('clean_4_muscle.csv')
bone = pd.read_csv('clean_5_bone.csv')
vitals = pd.read_csv('clean_6_vitals.csv')

# 2. Sample 5,000 synchronized rows across all 6 cleaned sources
n = 5000
np.random.seed(42)

v_s = vitals.sample(n=n, replace=True, random_state=42).reset_index(drop=True)
g_s = gluc.sample(n=n, replace=True, random_state=42).reset_index(drop=True)
sm_s = sleep_mood.sample(n=n, replace=True, random_state=42).reset_index(drop=True)
m_s = muscle.sample(n=n, replace=True, random_state=42).reset_index(drop=True)
b_s = bone.sample(n=n, replace=True, random_state=42).reset_index(drop=True)
r_s = rad.sample(n=n, replace=True, random_state=42).reset_index(drop=True)

# 3. Combine into your strict 9-indicator master dataset
master_df = pd.DataFrame({
    'systolic_bp': v_s['systolic_bp'],
    'diastolic_bp': v_s['diastolic_bp'],
    'heart_rate': v_s['heart_rate'],
    'blood_sugar': g_s['blood_sugar'],
    'body_temperature': v_s['body_temperature'],
    'daily_sleep_hours': sm_s['daily_sleep_hours'],
    'mood': sm_s['mood'],
    'muscle_strength_kg': m_s['muscle_strength_kg'],
    'bone_density_t_score': b_s['bone_density_t_score'],
    'radiation_mgy_day': r_s['radiation_mgy_day']
})

# Inject 10% realistic spaceflight emergency spikes (solar flares, tachycardia, severe bone loss)
spike_idx = np.random.choice(n, size=int(n * 0.10), replace=False)
master_df.loc[spike_idx, 'radiation_mgy_day'] = np.random.uniform(0.45, 2.50, len(spike_idx)).round(3)
master_df.loc[spike_idx[:250], 'heart_rate'] = np.random.uniform(118, 148, 250).round(1)
master_df.loc[spike_idx[250:], 'bone_density_t_score'] = np.random.uniform(-2.7, -2.0, len(spike_idx) - 250).round(2)

# 4. Assign Spaceflight Risk Labels (0 = Optimal, 1 = Warning, 2 = Critical)
def assign_risk(row):
    if (
        row['systolic_bp'] > 145 or row['systolic_bp'] < 85 or
        row['heart_rate'] > 115 or row['heart_rate'] < 45 or
        row['blood_sugar'] < 65 or row['blood_sugar'] > 155 or
        row['body_temperature'] > 38.0 or row['body_temperature'] < 35.5 or
        row['bone_density_t_score'] < -2.0 or
        row['radiation_mgy_day'] > 0.40
    ):
        return 2  # Critical
    elif (
        row['systolic_bp'] > 130 or
        row['heart_rate'] > 95 or
        row['daily_sleep_hours'] < 6.0 or
        row['mood'] <= 2 or
        row['muscle_strength_kg'] < 28.0 or
        row['bone_density_t_score'] < -1.0 or
        row['radiation_mgy_day'] > 0.25
    ):
        return 1  # Warning
    return 0      # Optimal

master_df['risk_level'] = master_df.apply(assign_risk, axis=1)

# 5. Scale features & split data
X = master_df.drop(columns=['risk_level'])
y = master_df['risk_level']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# 6. Train Model A: Unsupervised Autoencoder (learns normal healthy baseline)
X_train_normal = X_train[y_train == 0]
input_dim = X_train.shape[1]

inp = Input(shape=(input_dim,))
enc = Dense(8, activation='relu')(inp)
bot = Dense(4, activation='relu')(enc)
dec = Dense(8, activation='relu')(bot)
out = Dense(input_dim, activation='linear')(dec)

autoencoder = Model(inp, out)
autoencoder.compile(optimizer='adam', loss='mse')
autoencoder.fit(X_train_normal, X_train_normal, epochs=40, batch_size=32, verbose=0)

train_recon = autoencoder.predict(X_train_normal, verbose=0)
threshold = np.percentile(np.mean(np.square(X_train_normal - train_recon), axis=1), 95)

# 7. Train Model B: Supervised Random Forest Classifier
rf_model = RandomForestClassifier(n_estimators=150, random_state=42)
rf_model.fit(X_train, y_train)

print("Autoencoder Anomaly Threshold (MSE):", round(threshold, 4))
print("\nRandom Forest Risk Classification Report:\n")
print(classification_report(y_test, rf_model.predict(X_test)))

# 8. Save all trained models & scaler for download
autoencoder.save('astronaut_autoencoder.keras')
joblib.dump(rf_model, 'astronaut_rf_model.pkl')
joblib.dump(scaler, 'astronaut_scaler.pkl')
joblib.dump({'threshold': threshold, 'features': list(X.columns)}, 'astronaut_config.pkl')
master_df.to_csv('final_astronaut_9features.csv', index=False)
print("Saved all models and final_astronaut_9features.csv successfully!")
```

```python
import pandas as pd
import numpy as np
import joblib
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense

# 1. Load your 6 cleaned files
rad = pd.read_csv('clean_1_radiation.csv')
gluc = pd.read_csv('clean_2_glucose.csv')
sleep_mood = pd.read_csv('clean_3_sleep_mood.csv')
muscle = pd.read_csv('clean_4_muscle.csv')
bone = pd.read_csv('clean_5_bone.csv')
vitals = pd.read_csv('clean_6_vitals.csv')

# 2. Filter healthy pools first so astronauts start from a realistic healthy baseline
healthy_vitals = vitals[
    (vitals['systolic_bp'].between(95, 128)) &
    (vitals['diastolic_bp'].between(60, 82)) &
    (vitals['heart_rate'].between(55, 88)) &
    (vitals['body_temperature'].between(36.3, 37.3))
]
healthy_sleep = sleep_mood[(sleep_mood['daily_sleep_hours'] >= 6.2) & (sleep_mood['mood'] >= 3)]
healthy_muscle = muscle[muscle['muscle_strength_kg'] >= 30.0]
healthy_bone = bone[bone['bone_density_t_score'] >= -0.9]

np.random.seed(42)

# Build 3,000 Optimal (0), 1,200 Warning (1), and 800 Critical (2) astronaut records
def sample_pool(n_rows):
    return pd.DataFrame({
        'systolic_bp': healthy_vitals['systolic_bp'].sample(n_rows, replace=True).values.astype(float),
        'diastolic_bp': healthy_vitals['diastolic_bp'].sample(n_rows, replace=True).values.astype(float),
        'heart_rate': healthy_vitals['heart_rate'].sample(n_rows, replace=True).values.astype(float),
        'blood_sugar': gluc['blood_sugar'].sample(n_rows, replace=True).values.astype(float),
        'body_temperature': healthy_vitals['body_temperature'].sample(n_rows, replace=True).values.astype(float),
        'daily_sleep_hours': healthy_sleep['daily_sleep_hours'].sample(n_rows, replace=True).values.astype(float),
        'mood': healthy_sleep['mood'].sample(n_rows, replace=True).values.astype(int),
        'muscle_strength_kg': healthy_muscle['muscle_strength_kg'].sample(n_rows, replace=True).values.astype(float),
        'bone_density_t_score': healthy_bone['bone_density_t_score'].sample(n_rows, replace=True).values.astype(float),
        'radiation_mgy_day': rad['radiation_mgy_day'].sample(n_rows, replace=True).values.astype(float)
    })

# 3,000 Optimal Baseline Rows
df_opt = sample_pool(3000)

# 1,200 Warning Rows (mild spaceflight drift: sleep loss, mild bone/muscle dip, elevated radiation)
df_warn = sample_pool(1200)
df_warn.loc[:399, 'daily_sleep_hours'] = np.random.uniform(4.8, 5.8, 400).round(1)
df_warn.loc[:399, 'mood'] = np.random.choice([1, 2], 400)
df_warn.loc[400:799, 'bone_density_t_score'] = np.random.uniform(-1.9, -1.1, 400).round(2)
df_warn.loc[400:799, 'muscle_strength_kg'] = np.random.uniform(22.5, 27.5, 400).round(1)
df_warn.loc[800:, 'radiation_mgy_day'] = np.random.uniform(0.26, 0.39, 400).round(3)
df_warn.loc[800:, 'heart_rate'] = np.random.uniform(96, 112, 400).round(1)

# 800 Critical Rows (acute emergencies: solar flare, tachycardia, severe bone loss, hypoglycemia)
df_crit = sample_pool(800)
df_crit.loc[:299, 'radiation_mgy_day'] = np.random.uniform(0.45, 2.50, 300).round(3)
df_crit.loc[:299, 'heart_rate'] = np.random.uniform(118, 148, 300).round(1)
df_crit.loc[300:549, 'bone_density_t_score'] = np.random.uniform(-2.9, -2.05, 250).round(2)
df_crit.loc[550:, 'blood_sugar'] = np.random.uniform(48, 63, 250).round(1)
df_crit.loc[550:, 'body_temperature'] = np.random.uniform(38.1, 39.4, 250).round(2)

master_df = pd.concat([df_opt, df_warn, df_crit], ignore_index=True)

def assign_risk(row):
    if (
        row['systolic_bp'] > 145 or row['systolic_bp'] < 85 or
        row['heart_rate'] > 115 or row['heart_rate'] < 45 or
        row['blood_sugar'] < 65 or row['blood_sugar'] > 155 or
        row['body_temperature'] > 38.0 or row['body_temperature'] < 35.5 or
        row['bone_density_t_score'] < -2.0 or
        row['radiation_mgy_day'] > 0.40
    ):
        return 2
    elif (
        row['systolic_bp'] > 130 or
        row['heart_rate'] > 95 or
        row['daily_sleep_hours'] < 6.0 or
        row['mood'] <= 2 or
        row['muscle_strength_kg'] < 28.0 or
        row['bone_density_t_score'] < -1.0 or
        row['radiation_mgy_day'] > 0.25
    ):
        return 1
    return 0

master_df['risk_level'] = master_df.apply(assign_risk, axis=1)

# 3. Scale & Train Both Models
X = master_df.drop(columns=['risk_level'])
y = master_df['risk_level']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42, stratify=y)

X_train_normal = X_train[y_train == 0]
input_dim = X_train.shape[1]

inp = Input(shape=(input_dim,))
enc = Dense(8, activation='relu')(inp)
bot = Dense(4, activation='relu')(enc)
dec = Dense(8, activation='relu')(bot)
out = Dense(input_dim, activation='linear')(dec)

autoencoder = Model(inp, out)
autoencoder.compile(optimizer='adam', loss='mse')
autoencoder.fit(X_train_normal, X_train_normal, epochs=50, batch_size=32, verbose=0)

train_recon = autoencoder.predict(X_train_normal, verbose=0)
threshold = np.percentile(np.mean(np.square(X_train_normal - train_recon), axis=1), 95)

rf_model = RandomForestClassifier(n_estimators=150, random_state=42)
rf_model.fit(X_train, y_train)

print("New Autoencoder Anomaly Threshold (MSE):", round(threshold, 4))
print("\nBalanced Random Forest Classification Report:\n")
print(classification_report(y_test, rf_model.predict(X_test)))

# Overwrite saved models with the improved versions
autoencoder.save('astronaut_autoencoder.keras')
joblib.dump(rf_model, 'astronaut_rf_model.pkl')
joblib.dump(scaler, 'astronaut_scaler.pkl')
joblib.dump({'threshold': threshold, 'features': list(X.columns)}, 'astronaut_config.pkl')
master_df.to_csv('final_astronaut_9features.csv', index=False)
```
